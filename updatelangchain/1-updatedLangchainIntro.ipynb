# Langchain Version V1

```python
import os
from dotenv import load_dotenv
load_dotenv()

os.environ["ANTHROPIC_API_KEY"] = os.getenv("ANTHROPIC_API_KEY")
```

```python

from langchain.agents import create_agent

def get_weather(city: str) -> str:
    """Return the weather for a given city."""
    return f"The weather in {city} is sunny."

agent = create_agent(
    model="claude-sonnet-4-6",
    tools=[get_weather],
    system_prompt="You are a helpful assistant.",
)

result = agent.invoke({
    "messages": [
        {"role": "user", "content": "What is the weather in New York?"}
    ]
})

print(result["messages"][-1].content)
```

# Model integration with Anthropic


```python
from langchain.chat_models import init_chat_model
model = init_chat_model(model="claude-sonnet-4-6")
```

# Streaming and Batch

```python
response = model.stream("Write the 200 word paragrapth on the blockchain?")
for chunk in response:
    print(chunk.text, end="", flush=True) #Streaming basiclly use for the large text generation and it will print the text as it is generated.
```

## Batch


```python
response =model.batch([
    "Write a poem about the ocean.",
    "What is the capital of France?",
    "why do we need to learn programming?"
])
for r in response:
    print(r)
```

```python
response =model.batch([
    "Write a poem about the ocean.",
    "What is the capital of France?",
    "why do we need to learn programming?"
],
config={"max_concurrency": 5})
for r in response:
    print(r)
```
