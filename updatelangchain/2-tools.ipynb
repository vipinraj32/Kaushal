```python
import os
from dotenv import load_dotenv
from langchain.chat_models import init_chat_model
load_dotenv()
os.environ["ANTHROPIC_API_KEY"]=os.getenv("ANTHROPIC_API_KEY")

model=init_chat_model(model="claude-sonnet-4-6")
response=model.stream("Write the 200 word paragrapth on the blockchain?")
for chunk in response:
    print(chunk.text, end="", flush=True) #Streaming basiclly use for the large text generation and it will print the text as it is generated.

```

```python
from langchain.tools import tool

@tool
def get_weather(city: str) -> str:
    """Get the current weather in a given city."""
    return f"the weather in {city} is sunny"

model_with_tools = model.bind_tools([get_weather])
```

```python
response = model_with_tools.invoke(
    "What is the weather in New York City?"
)

print("AI Response:", response.content)

for tool_call in response.tool_calls:
    print(f"Tool: {tool_call['name']}")
    print(f"Input: {tool_call['args']}")
```

### Tool Execution loop

```python
# Step 1 : Model Generates tool cells
messages = [{"role": "user", "content": "What is the weather in New York City?"}]
response = model_with_tools.invoke(messages)
messages.append(response)

#Step 2 : Execute the tools and collect the results
for tool_call in response.tool_calls:
    #Execute the tool with the generated input
    tool_result = get_weather.invoke(tool_call)
    messages.append(tool_result)

# Step 3 : Pass the results back to the model for final response
final_response = model_with_tools.invoke(messages)
print("Final AI Response:", final_response.text)
```

```python
from langchain.messages import AIMessage, HumanMessage, SystemMessage, ToolMessage
from langchain.tools import tool

ai_msg=AIMessage(content=[],
                 tool_calls=[{"name": "get_weather", "args": {"location": "New York City"},
                              "id": "tool_call_1"}]) 

weather_result="Sunny, 72°F"

tool_msg=ToolMessage(content=weather_result, tool_name="get_weather", tool_call_id="tool_call_1")

messages=[
    HumanMessage(content="What is the weather in New York City?"),
    ai_msg,
    tool_msg
]

response=model.invoke(messages)
```
