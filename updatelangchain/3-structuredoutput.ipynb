## Message output alongside parsed structure

```python
from pydantic import BaseModel, Field

class Movie(BaseModel):
    title: str = Field(..., description="The title of the movie")
    year: int = Field(..., description="The release year of the movie")
    genre: str = Field(..., description="The genre of the movie")
    director: str = Field(..., description="The director of the movie")
    actors: list[str] = Field(..., description="A list of main actors in the movie")
    rating: float = Field(..., description="The rating of the movie (out of 10)")
model_with_structured_output = model.with_structured_output(Movie, include_raw=True)
response = model_with_structured_output.invoke("Provide details about the movie dhamal4, including its title, release year, genre, director, and rating.")
response
```

## Nested Structure

```python
from pydantic import BaseModel, Field

class Actor(BaseModel):
    name: str = Field(..., description="The name of the actor")
    birth_year: int = Field(..., description="The birth year of the actor")
    nationality: str = Field(..., description="The nationality of the actor")
    role: str = Field(..., description="The role played by the actor in the movie")

class MovieDetails(BaseModel):
    title: str = Field(..., description="The title of the movie")
    year: int = Field(..., description="The release year of the movie")
    genre: str = Field(..., description="The genre of the movie")
    director: str = Field(..., description="The director of the movie")
    actors: list[Actor] = Field(..., description="A list of main actors in the movie")
    rating: float = Field(..., description="The rating of the movie (out of 10)")

model_with_structured_output = model.with_structured_output(MovieDetails, include_raw=True)

response = model_with_structured_output.invoke("Provide details about the movie dhamal4, including its title, release year, genre, director, actors (with their birth year and nationality), and rating.")
response
```

```python
from typing_extensions import Annotated,TypedDict

class MovieDict(TypedDict):
    title: Annotated[str, "The title of the movie"]
    year: Annotated[int, "The release year of the movie"]
    genre: Annotated[str, "The genre of the movie"]
    director: Annotated[str, "The director of the movie"]
    actors: Annotated[list[dict[str, str]], "A list of All actors in the movie"]
    rating: Annotated[float, "The rating of the movie (out of 10)"]

model_with_structured_output = model.with_structured_output(MovieDict, include_raw=True)
response = model_with_structured_output.invoke("Provide details about the movie dhamal4, including its title, release year, genre, director, actors (with their birth year and nationality), and rating.")  
response
```

## DataClasses

```python
from langchain.agents import create_agent

class ContactInfo(BaseModel):
    name: str = Field(..., description="The name of the contact person")
    email: str = Field(..., description="The email address of the contact person")
    phone: str = Field(..., description="The phone number of the contact person")

agent = create_agent(model= "claude-sonnet-4-6",
                     response_format= ContactInfo)

response = agent.invoke({"messages":[{"role": "user", "content": "Provide contact information for John Doe, including name, email, and phone number."}]})
response
```

## Data Classes for Structured Output

```python
from dataclasses import dataclass
from langchain.agents import create_agent

@dataclass
class ContactInfo:
    name: str  # the Name of the contact person
    email: str # the email address of the contact person
    phone: str # the phone number of the contact person

agent = create_agent(model="claude-sonnet-4-6",
                     response_format=ContactInfo)

result=agent.invoke({"messages":[{"role": "user", "content": "Provide contact information for John Doe, including name, email, and phone number."}]})
result["structured_response"]
```
